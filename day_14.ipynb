Write a Python function to merge two sorted lists into a single sorted list.
Input: [1, 3, 5], [2, 4, 6]
Output: [1, 2, 3, 4, 5, 6]

```python
a = [1, 3, 5]
b = [2, 4, 6]

c = []

for x in range(3):
    if a[x] < b[x]:
        c += [a[x], b[x]]
    else:
        c += [b[x], a[x]]

print(c)
```

Write a Python function to reverse a given string.
Input: "hello"
Output: "olleh"

```python
word = 'python'

rev_word = ''

for x in word:
    rev_word = x + rev_word

print(rev_word)
```

Write a Python function to check if two given strings are anagrams of each other.
Input: "listen", "silent"
Output: True

```python
word1 = 'listen'
word2 = 'silent'

count = 0

for x in word1:
    if x in word2:
        count = count + 1

if count == len(word1):
    print(True)
else:
    print(False)
```
